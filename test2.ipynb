```python
import numpy as np

def row_reduce_to_systematic_form(G):
    # Get the number of rows and columns in G
    rows, cols = G.shape
    k = rows  # number of rows (information bits)

    # Gaussian elimination to put G in row echelon form
    for i in range(k):
        # Step 1: Make G[i][i] a 1 by row swapping if necessary
        if G[i, i] == 0:
            for j in range(i + 1, k):
                if G[j, i] != 0:
                    G[[i, j]] = G[[j, i]]  # Swap row i with row j
                    break
        
        # Step 2: Normalize the row to make the diagonal element G[i][i] equal to 1
        if G[i, i] != 1:
            G[i] = G[i] / G[i, i]
        
        # Step 3: Eliminate all other elements in column i
        for j in range(k):
            if i != j and G[j, i] != 0:
                G[j] = G[j] - G[j, i] * G[i]

    # Return the reduced matrix G in systematic form
    return G

# Example usage
G = np.array([
    [4, 3, 1, 4, 3],
    [3, 1, 2, 0, 4],
    [4, 1, 4, 2, 4]
], dtype=float)

print(G[1] / 3)

```
